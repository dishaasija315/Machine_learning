```python
# 1. Function to find square of a number
def square(n):
    return n * n

print(square(6))
```

```python
# 2. Function to check even or odd
def check_even_odd(num):
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")

check_even_odd(7)
```

```python
# 3. Function to find maximum of two numbers
def maximum(a, b):
    if a > b:
        return a
    else:
        return b

print(maximum(10, 25))
```

```python
# 4. Function to calculate factorial
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact = fact * i
    return fact

print(factorial(5))
```

```python
# 5. Function to reverse a string
def reverse_string(text):
    return text[::-1]

print(reverse_string("python"))
```

```python
# 6. Function to count vowels
def count_vowels(word):
    vowels = "aeiouAEIOU"
    count = 0

    for i in word:
        if i in vowels:
            count += 1

    print("Vowels =", count)

count_vowels("Artificial Intelligence")
```

```python
# 7. Function using lambda
square = lambda x: x*x

print(square(8))
```

```python
# 8. Recursive function
def countdown(n):
    if n == 0:
        print("Done")
    else:
        print(n)
        countdown(n-1)

countdown(5)
```

```python
# 9. Function to find sum of list elements
def list_sum(numbers):
    total = 0

    for i in numbers:
        total += i

    return total

print(list_sum([1,2,3,4,5]))
```

```python
# 10. Function with keyword arguments
def student(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student(course="CSIT", age=19, name="Disha")
```

```python
# 11. Function to check palindrome
def palindrome(word):
    if word == word[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")

palindrome("madam")
```

```python
# 12. Function to generate multiplication table
def table(n):
    for i in range(1,11):
        print(n, "x", i, "=", n*i)

table(5)
```

```python
# 15. Function to find largest element in list
def largest(lst):
    print(max(lst))

largest([12,45,2,89,34])
```

```python
# 16. Function to find prime number
def is_prime(n):
    if n <= 1:
        print("Not Prime")
        return

    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            return

    print("Prime")

is_prime(11)
```

```python
# 17. Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(c):
    f = (c * 9/5) + 32
    print("Fahrenheit =", f)

celsius_to_fahrenheit(37)
```

```python
# 22. Function to check leap year
def leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print("Leap Year")
    else:
        print("Not Leap Year")

leap_year(2024)
```

```python
# 23. Function to find average of numbers
def average(numbers):
    avg = sum(numbers) / len(numbers)
    print("Average =", avg)

average([10,20,30,40])
```

```python
# 24. Function to count uppercase and lowercase letters
def count_case(text):
    upper = 0
    lower = 0

    for i in text:
        if i.isupper():
            upper += 1
        elif i.islower():
            lower += 1

    print("Uppercase =", upper)
    print("Lowercase =", lower)

count_case("PyThOn")
```
